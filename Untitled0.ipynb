```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load the modulated data (이미지 데이터를 불러옵니다)
path = "/content/drive/MyDrive/modulated_data_D.csv"
modulated_data = pd.read_csv(path, header=None).values.flatten()

# Parameters
f = 10  # Frequency (Hz)
time_step = 0.01  # Time interval (s)
num_sample = int(1 / time_step)  # 샘플 수
t = np.arange(0, 1, time_step)  # Time vector for one bit (1 second duration)
row = 100  # Image row size
col = 1000  # Image column size

# 첫 번째 값이 0이면 제거
if modulated_data[0] == 0:
    modulated_data = modulated_data[1:]

# Generate carrier signal
t = np.arange(0, num_sample * time_step, time_step)
carrier_signal = np.cos(2 * np.pi * f * t)
carrier_signal_full = np.tile(carrier_signal, len(modulated_data) // num_sample)

# Add AWGN noise to the modulated signal
def add_noise(signal, sigma):
    noise = np.random.normal(0, sigma, signal.shape)
    return signal + noise

# Decoding function
def decode_bpsk(signal, carrier_signal, row, col):
    # Multiply by carrier signal
    demodulated_signal = signal * carrier_signal
    # Integrate over each period
    integrated_signal = np.sum(demodulated_signal.reshape(-1, num_sample), axis=1) / num_sample
    # Reshape into image dimensions
    reshaped_signal = integrated_signal.reshape(row, col)
    # Threshold decision rule
    decoded_data = (reshaped_signal > 0).astype(int)
    return decoded_data

# Q1: Plot at "A" (Modulated Signal)
plt.figure(figsize=(12, 4))
plt.title("Plot at A: Modulated Signal")
plt.plot(modulated_data[:500])  # First 500 samples
plt.xlabel("Sample Index")
plt.ylabel("Amplitude")
plt.grid()
plt.show()

# Q2: Plot at "B" (Demodulated Signal)
demodulated_signal = modulated_data * carrier_signal_full
plt.figure(figsize=(12, 4))
plt.title("Plot at B: Demodulated Signal")
plt.plot(demodulated_signal[:500])  # First 500 samples of demodulated signal
plt.xlabel("Sample Index")
plt.ylabel("Amplitude")
plt.grid()
plt.show()

# Q3: Image at "C" (Reshaped Image Before Threshold)
integrated_signal = np.sum(demodulated_signal.reshape(-1, num_sample), axis=1) / num_sample
reshaped_signal = integrated_signal.reshape(row, col)

plt.figure(figsize=(8, 8))
plt.title("Image at C: Reshaped Image (Before Threshold)")
plt.imshow(reshaped_signal, cmap='gray')
plt.show()

# Q5: Final Decoded Image at "D"
decoded_image_clean = decode_bpsk(modulated_data, carrier_signal_full, row, col)

plt.figure(figsize=(8, 8))
plt.title("Decoded Image (Noise-Free)")
plt.imshow(decoded_image_clean, cmap='gray')
plt.show()

# Q7: Fix sigma=4 and show the noisy decoded image
sigma = 4
noisy_modulated_data = add_noise(modulated_data, sigma)
decoded_image_noisy = decode_bpsk(noisy_modulated_data, carrier_signal_full, row, col)

plt.figure(figsize=(8, 8))
plt.title("Decoded Image with Noise (σ=4)")
plt.imshow(decoded_image_noisy, cmap='gray')
plt.show()

# Q8: Compute BER for σ=1 to σ=10 and plot BER vs log10(1/σ)
sigmas = range(1, 11)
ber_list = []

for sigma in sigmas:
    ber_sum = 0
    for _ in range(10):  # Repeat experiment 10 times for averaging
        noisy_modulated_data = add_noise(modulated_data, sigma)
        decoded_image_noisy = decode_bpsk(noisy_modulated_data, carrier_signal_full, row, col)
        errors = np.sum(decoded_image_noisy.flatten() != decoded_image_clean.flatten())
        ber_sum += errors / (row * col)  # Normalize by total number of bits
    ber_list.append(ber_sum / 10)

# Plot BER vs log10(1/σ)
log_inv_sigma = np.log10(1 / np.array(sigmas))
plt.figure(figsize=(8, 6))
plt.plot(log_inv_sigma, ber_list, marker='o')
plt.title("BER vs log10(1/σ)")
plt.xlabel("log10(1/σ)")
plt.ylabel("Bit Error Rate (BER)")
plt.grid()
plt.show()




```
